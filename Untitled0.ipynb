```python
%%writefile model.py
import torch
import torch.nn as nn


class Chomp1d(nn.Module):
    def __init__(self, chomp_size):
        super().__init__()
        self.chomp_size = chomp_size

    def forward(self, x):
        if self.chomp_size == 0:
            return x
        return x[:, :, :-self.chomp_size].contiguous()


class TemporalBlock(nn.Module):
    def __init__(self, in_channels, out_channels, kernel_size, dilation, dropout=0.3):
        super().__init__()
        padding = (kernel_size - 1) * dilation

        self.conv1 = nn.Conv1d(in_channels, out_channels, kernel_size,
                                padding=padding, dilation=dilation)
        self.chomp1 = Chomp1d(padding)
        self.bn1 = nn.BatchNorm1d(out_channels)
        self.relu1 = nn.ReLU()
        self.drop1 = nn.Dropout(dropout)

        self.conv2 = nn.Conv1d(out_channels, out_channels, kernel_size,
                                padding=padding, dilation=dilation)
        self.chomp2 = Chomp1d(padding)
        self.bn2 = nn.BatchNorm1d(out_channels)
        self.relu2 = nn.ReLU()
        self.drop2 = nn.Dropout(dropout)

        self.net = nn.Sequential(
            self.conv1, self.chomp1, self.bn1, self.relu1, self.drop1,
            self.conv2, self.chomp2, self.bn2, self.relu2, self.drop2,
        )

        if in_channels != out_channels:
            self.downsample = nn.Sequential(
                nn.Conv1d(in_channels, out_channels, 1),
                nn.BatchNorm1d(out_channels),
            )
        else:
            self.downsample = None

        self.relu_out = nn.ReLU()
        self._init_weights()

    def _init_weights(self):
        nn.init.kaiming_normal_(self.conv1.weight, nonlinearity="relu")
        nn.init.kaiming_normal_(self.conv2.weight, nonlinearity="relu")
        if self.downsample is not None:
            nn.init.kaiming_normal_(self.downsample[0].weight, nonlinearity="relu")

    def forward(self, x):
        out = self.net(x)
        res = x if self.downsample is None else self.downsample(x)
        return self.relu_out(out + res)


class ECG_TCN(nn.Module):
    def __init__(self, num_classes=5, seq_len=140, f1=2, ft=11, kt=11, dropout=0.3):
        super().__init__()
        self.input_conv = nn.Conv1d(1, f1, kernel_size=1)

        dilations = [1, 2, 4]
        blocks = []
        in_ch = f1
        for d in dilations:
            blocks.append(TemporalBlock(in_ch, ft, kernel_size=kt,
                                         dilation=d, dropout=dropout))
            in_ch = ft
        self.tcn_blocks = nn.Sequential(*blocks)

        self.flatten = nn.Flatten()
        self.fc = nn.Linear(ft * seq_len, num_classes)
        nn.init.kaiming_normal_(self.fc.weight, nonlinearity="relu")

    def forward(self, x):
        x = self.input_conv(x)
        x = self.tcn_blocks(x)
        x = self.flatten(x)
        return self.fc(x)


def count_parameters(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)


if __name__ == "__main__":
    model = ECG_TCN(num_classes=5)
    dummy = torch.randn(8, 1, 140)
    out = model(dummy)
    print("Output shape:", out.shape)
    print("Trainable parameters:", count_parameters(model))
```

```python
%%writefile data.py
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader


def _load_txt(path):
    data = np.loadtxt(path)
    y = data[:, 0].astype(np.int64)
    X = data[:, 1:].astype(np.float32)
    return X, y


def load_ecg5000(data_dir):
    X_train, y_train = _load_txt(f"{data_dir}/ECG5000_TRAIN.txt")
    X_test, y_test = _load_txt(f"{data_dir}/ECG5000_TEST.txt")

    y_train = y_train - 1
    y_test = y_test - 1

    def normalize(X):
        mean = X.mean(axis=1, keepdims=True)
        std = X.std(axis=1, keepdims=True) + 1e-8
        return (X - mean) / std

    X_train = normalize(X_train)
    X_test = normalize(X_test)

    return X_train, y_train, X_test, y_test


class ECG5000Dataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(X, dtype=torch.float32).unsqueeze(1)
        self.y = torch.tensor(y, dtype=torch.long)

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]


def get_dataloaders(data_dir, batch_size=30):
    X_train, y_train, X_test, y_test = load_ecg5000(data_dir)
    train_ds = ECG5000Dataset(X_train, y_train)
    test_ds = ECG5000Dataset(X_test, y_test)
    train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True)
    test_loader = DataLoader(test_ds, batch_size=batch_size, shuffle=False)
    return train_loader, test_loader
```

```python
%%writefile train.py
import argparse
import torch
import torch.nn as nn
import numpy as np
from sklearn.metrics import (
    accuracy_score, balanced_accuracy_score, confusion_matrix,
    precision_score, recall_score, f1_score,
)

from model import ECG_TCN, count_parameters
from data import get_dataloaders


def evaluate_all_metrics(y_true, y_pred):
    return {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Balanced Acc.": balanced_accuracy_score(y_true, y_pred),
        "Precision (macro)": precision_score(y_true, y_pred, average="macro", zero_division=0),
        "Recall (macro)": recall_score(y_true, y_pred, average="macro", zero_division=0),
        "F1 (macro)": f1_score(y_true, y_pred, average="macro", zero_division=0),
        "F1 (weighted)": f1_score(y_true, y_pred, average="weighted", zero_division=0),
    }


def train_one_epoch(model, loader, optimizer, criterion, device):
    model.train()
    total_loss = 0
    for X, y in loader:
        X, y = X.to(device), y.to(device)
        optimizer.zero_grad()
        out = model(X)
        loss = criterion(out, y)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * X.size(0)
    return total_loss / len(loader.dataset)


@torch.no_grad()
def get_predictions(model, loader, device):
    model.eval()
    all_preds, all_labels = [], []
    for X, y in loader:
        X = X.to(device)
        out = model(X)
        preds = out.argmax(dim=1).cpu().numpy()
        all_preds.extend(preds)
        all_labels.extend(y.numpy())
    return np.array(all_labels), np.array(all_preds)


def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--data_dir", type=str, required=True)
    parser.add_argument("--epochs", type=int, default=20)
    parser.add_argument("--batch_size", type=int, default=30)
    parser.add_argument("--lr", type=float, default=1e-3)
    args = parser.parse_args()

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"Using device: {device}")

    train_loader, test_loader = get_dataloaders(args.data_dir, args.batch_size)

    model = ECG_TCN(num_classes=5).to(device)
    print(f"Trainable parameters: {count_parameters(model):,}")

    all_train_labels = []
    for _, y in train_loader:
        all_train_labels.extend(y.numpy())
    all_train_labels = np.array(all_train_labels)
    class_counts = np.bincount(all_train_labels, minlength=5)
    print(f"Training class counts: {class_counts}")
    class_weights = 1.0 / np.maximum(class_counts, 1)
    class_weights = class_weights / class_weights.sum() * 5
    class_weights = torch.tensor(class_weights, dtype=torch.float32).to(device)

    criterion = nn.CrossEntropyLoss(weight=class_weights)
    optimizer = torch.optim.Adam(model.parameters(), lr=args.lr)

    best_bal_acc = 0
    best_state = None
    for epoch in range(1, args.epochs + 1):
        train_loss = train_one_epoch(model, train_loader, optimizer, criterion, device)
        y_true, y_pred = get_predictions(model, test_loader, device)
        acc = accuracy_score(y_true, y_pred)
        bal_acc = balanced_accuracy_score(y_true, y_pred)
        if bal_acc > best_bal_acc:
            best_bal_acc = bal_acc
            best_state = {k: v.clone() for k, v in model.state_dict().items()}
        print(f"Epoch {epoch:3d} | train_loss={train_loss:.4f} "
              f"| test_acc={acc:.4f} | test_balanced_acc={bal_acc:.4f}")

    if best_state is not None:
        model.load_state_dict(best_state)
    torch.save(model.state_dict(), "ecg_tcn_best.pt")
    print("Saved trained model to ecg_tcn_best.pt")
    y_true, y_pred = get_predictions(model, test_loader, device)
    metrics = evaluate_all_metrics(y_true, y_pred)
    cm = confusion_matrix(y_true, y_pred)

    print("\n=== Final results (best balanced-accuracy checkpoint) ===")
    print(f"Accuracy:           {metrics['Accuracy']:.4f}  (paper reports 94.2%)")
    print(f"Balanced accuracy:  {metrics['Balanced Acc.']:.4f}  (paper reports 89.0%)")
    print(f"Precision (macro):  {metrics['Precision (macro)']:.4f}  (not reported by paper)")
    print(f"Recall (macro):     {metrics['Recall (macro)']:.4f}  (not reported by paper)")
    print(f"F1 (macro):         {metrics['F1 (macro)']:.4f}  (not reported by paper)")
    print(f"F1 (weighted):      {metrics['F1 (weighted)']:.4f}  (not reported by paper)")
    print("Confusion matrix:\n", cm)


if __name__ == "__main__":
    main()
```

```python
%%writefile predict.py
import argparse
import numpy as np
import torch

from model import ECG_TCN

CLASS_NAMES = {
    0: "Normal",
    1: "R-on-T Premature Ventricular Contraction",
    2: "Supraventricular Premature or Ectopic Beat",
    3: "Premature Ventricular Contraction",
    4: "Unclassifiable Beat",
}


def load_model(weights_path, device):
    model = ECG_TCN(num_classes=5)
    model.load_state_dict(torch.load(weights_path, map_location=device))
    model.to(device)
    model.eval()
    return model


def preprocess_single(raw_signal):
    x = np.asarray(raw_signal, dtype=np.float32)
    mean = x.mean()
    std = x.std() + 1e-8
    x = (x - mean) / std
    x = torch.tensor(x).view(1, 1, -1)
    return x


@torch.no_grad()
def predict_arrhythmia(raw_signal, model, device):
    x = preprocess_single(raw_signal).to(device)
    logits = model(x)
    probs = torch.softmax(logits, dim=1).cpu().numpy()[0]
    pred_class = int(probs.argmax())
    return {
        "predicted_class": pred_class,
        "predicted_label": CLASS_NAMES[pred_class],
        "confidence": float(probs[pred_class]),
        "all_probabilities": {CLASS_NAMES[i]: float(p) for i, p in enumerate(probs)},
    }


def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--weights", type=str, default="ecg_tcn_best.pt")
    parser.add_argument("--sample_file", type=str, required=True)
    args = parser.parse_args()

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model = load_model(args.weights, device)

    raw_signal = np.loadtxt(args.sample_file, delimiter=",")
    if raw_signal.ndim > 1:
        raw_signal = raw_signal[0]

    result = predict_arrhythmia(raw_signal, model, device)

    print("=== Single-sample prediction ===")
    print(f"Predicted class : {result['predicted_class']} ({result['predicted_label']})")
    print(f"Confidence      : {result['confidence']*100:.1f}%")
    print("Full probability breakdown:")
    for label, p in result["all_probabilities"].items():
        print(f"  {label:<45} {p*100:5.1f}%")


if __name__ == "__main__":
    main()
```

```python
%%writefile app.py
import numpy as np
import torch
import gradio as gr
import matplotlib.pyplot as plt

from model import ECG_TCN
from preprocess_raw_ecg import preprocess_raw_ecg

CLASS_NAMES = {
    0: "Normal",
    1: "R-on-T Premature Ventricular Contraction",
    2: "Supraventricular Premature or Ectopic Beat",
    3: "Premature Ventricular Contraction",
    4: "Unclassifiable Beat",
}
RISK_LEVEL = {0: "low", 1: "high", 2: "moderate", 3: "high", 4: "moderate"}
RISK_EMOJI = {"low": "🟢", "moderate": "🟡", "high": "🔴"}

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = ECG_TCN(num_classes=5)
model.load_state_dict(torch.load("ecg_tcn_best.pt", map_location=device))
model.to(device)
model.eval()


def normalize(beat):
    mean, std = beat.mean(), beat.std() + 1e-8
    return (beat - mean) / std


@torch.no_grad()
def classify_beat(beat_140):
    x = torch.tensor(normalize(beat_140), dtype=torch.float32).view(1, 1, -1).to(device)
    logits = model(x)
    probs = torch.softmax(logits, dim=1).cpu().numpy()[0]
    pred_class = int(probs.argmax())
    return pred_class, float(probs[pred_class]), probs


def verdict_markdown(pred_class, confidence):
    risk = RISK_LEVEL[pred_class]
    md = f"### {RISK_EMOJI[risk]} {CLASS_NAMES[pred_class]}\n**Confidence:** {confidence*100:.1f}%  |  **Risk level:** {risk.title()}"
    if confidence < 0.6:
        md += "\n> ⚠️ Low model confidence — recommend manual clinical review."
    return md


def predict_auto(file_obj, sampling_rate):
    if file_obj is None:
        return "Please upload a file first.", None, None, ""

    raw = np.loadtxt(file_obj.name, delimiter=",")
    raw = np.asarray(raw).flatten()

    # Case 1: already a single pre-segmented 140-value heartbeat
    if raw.shape[0] == 140:
        pred_class, confidence, probs = classify_beat(raw)
        prob_dict = {CLASS_NAMES[i]: float(p) for i, p in enumerate(probs)}

        fig, ax = plt.subplots(figsize=(7, 2.8))
        ax.plot(raw, color="#2563eb", linewidth=1.6)
        ax.set_title("Input heartbeat (140 samples, pre-segmented)", fontsize=11)
        ax.grid(alpha=0.25)
        fig.tight_layout()

        return verdict_markdown(pred_class, confidence), prob_dict, fig, "Detected: single pre-segmented beat."

    # Case 2: longer raw recording -> auto-segment it
    try:
        processed_beats = preprocess_raw_ecg(raw, sampling_rate=float(sampling_rate))
    except Exception as e:
        return f"⚠️ Error during preprocessing: {e}", None, None, ""

    processed_beats = processed_beats[:20]  # cap for speed/readability

    results = []
    class_counts = {name: 0 for name in CLASS_NAMES.values()}
    for beat in processed_beats:
        pred_class, confidence, _ = classify_beat(beat)
        results.append((pred_class, confidence))
        class_counts[CLASS_NAMES[pred_class]] += 1

    fig, ax = plt.subplots(figsize=(7, 2.8))
    ax.plot(raw[:int(sampling_rate * 5)], color="#2563eb", linewidth=1.0)
    ax.set_title(f"First 5 seconds of raw recording ({len(raw)} samples @ {sampling_rate}Hz)", fontsize=10)
    ax.grid(alpha=0.25)
    fig.tight_layout()

    overall_md = f"### Detected {len(processed_beats)} heartbeat(s) in raw recording\n\n"
    for name, count in class_counts.items():
        if count > 0:
            overall_md += f"- **{name}**: {count}\n"

    table_md = "**Per-beat breakdown:**\n\n| # | Prediction | Confidence |\n|---|---|---|\n"
    for i, (pred_class, confidence) in enumerate(results):
        table_md += f"| {i} | {CLASS_NAMES[pred_class]} | {confidence*100:.1f}% |\n"

    return overall_md, None, fig, table_md


custom_css = """
.gradio-container {max-width: 950px !important; margin: auto !important;}
#title_block {text-align: center; padding: 10px 0 0 0;}
#title_block h1 {font-size: 2rem; margin-bottom: 0.2rem;}
#title_block p {color: #6b7280; font-size: 0.95rem;}
footer {visibility: hidden}
"""

with gr.Blocks(css=custom_css, theme=gr.themes.Soft(primary_hue="blue")) as demo:
    with gr.Column(elem_id="title_block"):
        gr.Markdown("# 🫀 ECG Arrhythmia Detector")
        gr.Markdown(
            "Upload **either** a single pre-segmented 140-value heartbeat (e.g. from ECG5000) "
            "**or** a raw, continuous ECG recording of any length from any source — the system detects which one you gave it."
        )

    with gr.Row():
        with gr.Column(scale=1):
            file_input = gr.File(label="Upload ECG file (CSV/TXT)", file_types=[".csv", ".txt"])
            sr_input = gr.Number(label="Sampling rate (Hz) — only used for raw recordings", value=360)
            submit_btn = gr.Button("Analyze", variant="primary")

        with gr.Column(scale=1):
            verdict = gr.Markdown("### Awaiting input...")
            probs_out = gr.Label(label="Class probabilities (single-beat mode only)")

    plot = gr.Plot(label="Waveform")
    extra_table = gr.Markdown()

    submit_btn.click(
        fn=predict_auto,
        inputs=[file_input, sr_input],
        outputs=[verdict, probs_out, plot, extra_table],
    )

    gr.Markdown(
        "<div style='text-align:center; color:#9ca3af; font-size:0.8rem; margin-top:20px;'>"
        "Research prototype — not a certified medical device. Trained on ECG5000; accuracy on non-ECG5000 sources is not guaranteed."
        "</div>"
    )

if __name__ == "__main__":
    demo.launch(share=True)
```

```python
%%writefile preprocess_raw_ecg.py
import numpy as np
import neurokit2 as nk
from scipy.signal import resample


def preprocess_raw_ecg(raw_signal, sampling_rate, target_len=140, window_before=0.3, window_after=0.3):
    cleaned = nk.ecg_clean(raw_signal, sampling_rate=sampling_rate)

    _, info = nk.ecg_peaks(cleaned, sampling_rate=sampling_rate)
    r_peaks = info["ECG_R_Peaks"]

    samples_before = int(window_before * sampling_rate)
    samples_after = int(window_after * sampling_rate)

    beats = []
    for peak in r_peaks:
        start = peak - samples_before
        end = peak + samples_after
        if start < 0 or end > len(cleaned):
            continue

        beat = cleaned[start:end]
        beat_resampled = resample(beat, target_len)

        mean, std = beat_resampled.mean(), beat_resampled.std() + 1e-8
        beat_normalized = (beat_resampled - mean) / std

        beats.append(beat_normalized)

    if len(beats) == 0:
        raise ValueError("No valid heartbeats detected — check signal quality or sampling_rate.")

    return np.array(beats, dtype=np.float32)
```
